## 1. Định nghĩa bài toán quy hoạch tuyến tính


- Dạng chuẩn tắc của bài toán quy hoạch tuyến tính:
  - Hàm mục tiêu: 
  $$\text{min } c^T x\  \text{or max } c^T x$$
  - Ràng buộc: 
  $$Ax \geq b\ \text{or } Ax \leq b$$
  - Biến quyết định: 
  $$x \geq 0$$

- Dạng chính tắc của bài toán quy hoạch tuyến tính:
  - Hàm mục tiêu: 
  $$\text{min } c^T x$$
  - Ràng buộc: 
  $$Ax = b$$
  - Biến quyết định: 
  $$x \geq 0$$

### Đưa bài toán dạng tổng quát về dạng chuẩn tắc:
  - Nếu có ràng buộc dạng $Ax \leq b$, ta thêm biến phụ $s$ để chuyển thành $Ax + s = b$ với $s \geq 0$. (hiểu đơn giản là ta thêm một biến phụ để "bù" vào phần thiếu của ràng buộc)
  - Nếu có ràng buộc dạng $Ax \geq b$, ta thêm biến phụ $s$ để chuyển thành $Ax - s = b$ với $s \geq 0$. (hiểu đơn giản là ta thêm một biến phụ để "trừ" vào phần thừa của ràng buộc)
  - Nếu hàm mục tiêu là max, ta có thể chuyển thành min bằng cách nhân với -1 (đổi dấu): 
  $$\text{max } c^T x \equiv \text{min } (-c^T x)$$
  - Nếu có biến không âm, ta giữ nguyên. Nếu có biến không bị ràng buộc về dấu, ta có thể tách thành hai biến:
  $$x_i = u_i - v_i \ \text{với } u_i, v_i \geq 0$$

> ví dụ:

**a)**
$$f(x) = x_1 - x_2 + 2x_3 \rightarrow \max$$
$$
\begin{cases} 
x_1 - 2x_2 + 2x_3 = 1 \\ 
x_1 + x_2 - 3x_3 \le 4 \\ 
x_1 - 3x_2 + x_3 \ge 3 \\ 
x_1, x_2, x_3 \ge 0 
\end{cases}
$$

**b)**
$$f(x) = x_1 + 3x_2 + 4x_3 \rightarrow \min$$
$$
\begin{cases} 
x_1 + 2x_2 + x_3 \le 5 \\ 
2x_1 + 3x_2 + x_3 \ge 6 \\ 
x_2 \ge 0, x_3 \ge 0 
\end{cases}
$$

> lời giải:

**a)**
- Hàm mục tiêu:
$$\text{max } (x_1 - x_2 + 2x_3) \equiv \text{min } (-x_1 + x_2 - 2x_3)$$
- Ràng buộc:
$$\begin{cases} 
x_1 - 2x_2 + 2x_3 = 1 \\
x_1 + x_2 - 3x_3 + x_4 = 4 \\
x_1 - 3x_2 + x_3 - x_5 = 3 \\
x_i \ge 0 \text{ với } i = 1, 2, 3, 4, 5
\end{cases}
$$

**b)**
- Hàm mục tiêu:
$$\text{min } (x_1 + 3x_2 + 4x_3)$$
- Ràng buộc:
$$\begin{cases}
u_1 - v_1 + 2x_2 + x_3 + x_4 = 5 \\
2(u_1 - v_1) + 3x_2 + x_3 - x_5 = 6 \\
x_1 = u_1 - v_1 \text{ với } u_1, v_1 \ge 0 \\
x_2 \ge 0, x_3 \ge 0, x_4 \ge 0, x_5 \ge 0
\end{cases}

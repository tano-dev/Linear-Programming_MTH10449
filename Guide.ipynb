# Tổng kết hoàn thiện thuật toán Simplex

Chương trình giải bài toán quy hoạch tuyến tính (QHTT) theo phương pháp Simplex (2 pha) đã được hoàn thành tại [simplex.py](file:///d:/Linear-Programming_MTH10449/simplex.py). Chương trình hỗ trợ đầy đủ các yêu cầu mà bạn đưa ra và in chi tiết cấu trúc *từ vựng* sau mỗi bước lặp một cách trực quan.

## Những chức năng đã được xây dựng:

1. **Chuẩn hóa tự động:**
   - Hỗ trợ cả 2 dạng hàm mục tiêu (Min/Max).
   - Hỗ trợ khai báo ràng buộc tùy ý: `<=`, `>=`, `=`.
   - Hỗ trợ khai báo giới hạn biến linh hoạt: biến không âm (`>= 0`), biến không dương (`<= 0`) và biến tự do (tuỳ ý - unrestricted).
2. **Phương pháp 2 pha chuẩn xác:**
   - Hệ thống tự động thêm **biến bù** ($s$), **biến thặng dư** (cũng kí hiệu là $s$), và **biến giả** ($a$).
   - Nếu bài toán yêu cầu các biến giả (chứa ràng buộc `>=` hoặc `=`), hàm tự động chạy **Pha 1** (tối ưu hóa hàm $W$).
   - Tự động kiểm tra bài toán có vô nghiệm (Infeasible) sau khi kết thúc Pha 1 hay không.
3. **Từ vựng toán học trực quan (Dictionary Formatting):**
   - Chức năng in ra các từ vựng rất rõ ràng và dễ đọc ở định dạng: `Biến Cơ Sở = Hệ số - (hệ số) * Biến Phi Cơ Sở`
   - Các biến giả sẽ bị ẩn trong danh sách hiển thị của Pha 2 để làm sạch cấu trúc biểu thức.
   - Hàm mục tiêu ($W$ cho Pha 1 và $Z$ cho Pha 2) sẽ được đặt ở cuối mỗi từ vựng. Đối với bài toán Min, $Z$ được điều chỉnh tự động để đảm bảo giá trị in ra khớp với giá trị $Z$ thực tế (dù bên trong thuật toán đang tìm Max của $-Z$).
4. **Quy tắc Bland (Bland's Rule) chống xoay vòng:**
   - Thuật toán ưu tiên biến (cả biến vào và biến ra) có chỉ số nhỏ nhất khi có các tỉ số hoặc giá trị hệ số hòa nhau để đảm bảo chương trình không bao giờ bị lặp vô hạn.

```python
# Cấu hình bài toán
c = [1, 3, 4]                           # Hàm mục tiêu
A = [[1, 2, 1], [2, 3, 1]]              # Hệ số vế trái
b = [5, 6]                              # Hệ số vế phải
types = ['<=', '>=']                    # Loại ràng buộc
bounds = [(None, None), (0, None), (0, None)] # Ràng buộc của biến x1, x2, x3

# Khởi tạo thuật toán và chạy
solver = SimplexSolver(c, maximize=False, A=A, b=b, constraint_types=types, bounds=bounds)
x_opt, z_opt = solver.solve()
```
